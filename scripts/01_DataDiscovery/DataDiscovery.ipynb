### 2. Expansão da rede com YouTube Data Tools

Os IDs dos canais iniciais foram inseridos no módulo "Channel Network" (https://ytdt.digitalmethods.net/mod_channels_net.php) da ferramenta YouTube Data Tools para a geração da rede de canais conectados.

A rede é salva em formato .gdf que é processado para extração dos IDs dos canais. Após isso os IDs são submetidos ao módulo "Channel List" para extração dos metadados dos canais.

```python
# EXTRAIR OS IDS DOS CANAIS DO ARQUIVO GDF E SALVAR EM TXT E CSV

import os
import re
import pandas as pd

regex_patern = re.compile(r'UC[\w-]{22}')

input_files = [
	os.path.join(os.getcwd(), 'channelnet_seeds49_nodes55019_2025_12_30-01_59_15.gdf'),
	os.path.join(os.getcwd(), 'channelnet_seeds49_nodes1739_2025_12_29-12_11_22.gdf')
]

channel_ids = []
for input_file in input_files:
    with open(input_file, 'r', encoding='utf-8') as file:
        lines = file.readlines()
    for line in lines:
        matches = regex_patern.findall(line)
        channel_ids.extend(matches)
channel_ids = list(set(channel_ids))
print(f'Foram encontrados {len(channel_ids)} canais únicos no arquivo GDF.')

df_channels = pd.DataFrame(channel_ids, columns=['channel_id'])
output_csv = os.path.join(os.getcwd(), 'channel_ids.csv')
df_channels.to_csv(output_csv, index=False)

output_txt = os.path.join(os.getcwd(), 'channel_ids.txt')
with open(output_txt, 'w', encoding='utf-8') as file:
    file.write(','.join(channel_ids))
```

### 4. Processamento dos arquivos obtidos pela pesquisa

```python
# Juntar os arquivos e eliminar os duplicados

import os
import pandas as pd

buscas_arquivos = {
	'programação': '[programação]channelsearch_channels396_2025_12_29-11_01_21.csv',
	'programming': '[programming]channelsearch_channels498_2025_12_29-11_04_11.csv',
	'Python' : '[Python]channelsearch_channels480_2025_12_29-11_05_14.csv',
	'Java' : '[Java]channelsearch_channels459_2025_12_29-11_06_10.csv',
	'JavaScript' : '[JavaScript]channelsearch_channels224_2025_12_29-11_06_55.csv',
	'C/C++' : '[C_C++]channelsearch_channels496_2025_12_29-11_08_32.csv',
	'C#' : '[C#]channelsearch_channels203_2025_12_29-11_09_28.csv',
	'TODOS_DA_REDE' : 'channelsearch_channels55022_2025_12_30-12_11_01.csv'
}

# Criar dataframe com os arquivos CSV e adicionar coluna 'busca'
df = pd.DataFrame()
for busca, arquivo in buscas_arquivos.items():
	df_temp = pd.read_csv(os.path.join(os.getcwd(), arquivo), sep=',', encoding='utf-8')
	df_temp['busca'] = busca
	df = pd.concat([df, df_temp], ignore_index=True)

print(f'Quantidade de canais encontrados: {len(df)}')

# Remover linhas duplicadas com base no id do canal
df.drop_duplicates(subset=['id'], inplace=True)
df.reset_index(drop=True, inplace=True)

print(f'Quantidade de canais após remoção de duplicados: {len(df)}')
```

### 5. Aplicação dos critérios para seleção dos canais mais relevantes:
- Canais que possuem a palavra programação na descrição;
- Canais com mais de 10 mil inscritos, mais de 1 milhão de visualizações e mais de 450 vídeos;
- Canais que possuem país definido;
- Canais Brasileiros;

Arquivo salvo em formato JSON

```python
# FUNÇÃO PARA OBTER O ARROBA DOS CANAIS

import requests
from bs4 import BeautifulSoup
import json

def get_channel_info(channel_id):
	url = f"https://www.youtube.com/channel/{channel_id}/about"
	response = requests.get(url)
	response.raise_for_status()
	if response.status_code == 200:
		soup = BeautifulSoup(response.content, 'html.parser')
		try:
			script_ytInitialData = soup.find(lambda tag: tag.name == 'script' and 'var ytInitialData = ' in tag.text).text.split('var ytInitialData = ')[1].rstrip(';')
			json_data = json.loads(script_ytInitialData)
			arroba = json_data.get('metadata', {}).get('channelMetadataRenderer', {}).get('vanityChannelUrl', None).split('/')[-1] if json_data.get('metadata', {}).get('channelMetadataRenderer', {}).get('vanityChannelUrl', None) else None
			return arroba
		except Exception as err:
			print(f'Erro ao extrair script_ytInitialData: {err}')
			return None
```

```python
# APLICAR OS CRITÉRIOS DE INCLUSÃO

dados = df.copy()

dados = dados[dados['description'].notnull()] # Filtrar canais que possuem descrição
dados = dados[dados['description'].str.contains(r'(?<!\w)(?:programação|programming|coding|dev|python|java|javascript|c|c\/c\+\+|c\+\+|c#|csharp)(?!\w)', case=False, regex=True)] # Filtrar canais que possuem as palavras listadas programação na descrição
canais_filtrados = dados[(dados['viewCount'] > 1000000) & (dados['subscriberCount'] > 7000) & (dados['videoCount'] > 250)] # Filtrar canais com mais de 1 milhão de visualizações, 7 mil inscritos e 250 vídeos
canais_filtrados = canais_filtrados[canais_filtrados['country'] == 'BR'] # Filtrar canais brasileiros
canais_filtrados = canais_filtrados.sort_values(by=['viewCount', 'subscriberCount', 'videoCount'], ascending=False) # Ordenar canais
canais_filtrados['url'] = 'https://www.youtube.com/channel/' + canais_filtrados['id'] # Adicionar coluna com o link do canal
canais_filtrados['arroba'] = canais_filtrados['id'].apply(get_channel_info) # Adicionar coluna com o arroba do canal

canais_filtrados = canais_filtrados[[
    "id",
    "title",
    "arroba",
    "country",
    "description",
    "keywords",
    "viewCount",
    "subscriberCount",
    "videoCount",
    "url"
]]

print(f"Total de canais filtrados: {len(canais_filtrados)}")

canais_filtrados.to_json('canais_filtrados_DataTools.json', orient='records', force_ascii=False, indent=4)
canais_filtrados
```

### 6. Filtro manual

Após os processamentos, a lista de canais passa por um filtro feito manualmente para aplicar o critério qualitativo de permanência, resultando na lista de 80 canais.

```python
# Criar tabela csv com os canais selecionados

with open('canais_selecionados_DataTools.json', 'r', encoding='utf-8') as file:
	selecao_canais_final = json.load(file)

tabela_csv = pd.DataFrame(selecao_canais_final)
tabela_csv = tabela_csv[['id', 'title', 'arroba', 'description', 'keywords', 'url']]
tabela_csv.to_csv('canais_selecionados_DataTools.csv', index=False)

tabela_csv
```
